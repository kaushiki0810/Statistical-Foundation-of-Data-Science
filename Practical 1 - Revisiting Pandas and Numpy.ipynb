**Practical 1 - Revisiting Pandas and Numpy**


**Step-1** : Importing the necessary libraries.
             Pandas (pd) is used to create and manage DataFrames.
             NumPy (np) is used for numerical and matrix operations such as addition, multiplication, transpose, and inverse.

```python
import pandas as pd
import numpy as np
```

**Step-2**: Create a dataset.data stores the raw numerical values in rows and columns. pd.DataFrame() converts the raw data into a DataFrame. columns=["A", "B", "C"] assigns names to the three columns.

```python
data = [
    [2, 4, 5],
    [6, 8, 9],
    [7, 5, 6],

]

df = pd.DataFrame(data, columns=["A", "B", "C"])

print("DataFrame:")
print(df)
```

**Step 3**: Convert DataFrame to NumPy Array. df.to_numpy() converts the Pandas DataFrame into a NumPy array.
The array is stored in the variable arr.

```python
arr = df.to_numpy()

print("\nNumPy Array:")
print(arr)
```

**Step 4**: Finding Shape, Size and Dimension. arr.shape → shows the number of rows and columns.
arr.size → shows the total number of elements.
arr.ndim → shows the number of dimensions.

```python
print("\nShape:", arr.shape)
print("Size:", arr.size)
print("Dimension:", arr.ndim)
```

**Step-5**: Performing operations using numpy.


**1. Addition:-** arr + arr performs element-wise addition of the array with itself.
The result is stored in the variable addition.

```python
addition = arr + arr

print("\nAddition:")
print(addition)
```

**2. Subtraction:-** arr - arr performs element-wise subtraction.The result is stored in variable subtraction.

```python
subtraction = arr - arr

print("\nSubtraction:")
print(subtraction)
```

**3. Multiplication:-** Each value is multiplied by the corresponding value in the same array.

```python
multiplication = arr * arr

print("\nMultiplication:")
print(multiplication)
```

**4. Transpose:-** Transpose flips the rows and columns of the matrix.arr.T calculates the transpose of the array.

```python
transpose = arr.T

print("\nTranspose:")
print(transpose)
```

**5. Inverse:-** linalg → NumPy's linear algebra functions.
inv() → finds the inverse

```python
inverse = np.linalg.inv(arr)

print("Inverse:")
print(inverse)
```

**6. Product:-** np.dot(arr, arr) performs matrix multiplication of the array with itself.
The result is stored in the variable product.

```python
product = np.dot(arr, arr)

print("\nMatrix Product:")
print(product)
```
