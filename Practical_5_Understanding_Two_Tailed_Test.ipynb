**Q3 – Two-Tailed Z-Test for Mean Points Scored**

Given data-

Parameter        ----------------------------    Value

---
Historic mean (μ)     ----------------------   12 points

Population standard deviation (σ)  ------   5.5 points

Sample size (n)-------------------------36

Sample mean (x̄)-----------------------10.7 points

Significance level (α)-------------------0.05

**Step 1: State the hypothesis**

Null hypothesis (H₀): The mean points scored by regional players is not different from the historic mean of 12 points.

H0​:μ=12

Alternative hypothesis (H₁): The mean points scored by regional players is different from the historic mean of 12 points.

H1​:μ=12

This is a two-tailed test because we are checking for a difference in either direction.

**Step 2: Calculate the Z-score**


```python
#Code-
import math
from scipy.stats import norm

# Given values
mu = 12
sigma = 5.5
n = 36
sample_mean = 10.7
alpha = 0.05

# Calculate Z-score
z_score = (sample_mean - mu) / (sigma / math.sqrt(n))

# Calculate two-tailed p-value
p_value = 2 * norm.cdf(-abs(z_score))

print("Z-score:", z_score)
print("Two-tailed p-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

**Conclusion**

At the 5% significance level, the two-tailed test does not provide sufficient evidence to conclude that the regional players' mean scoring average differs from the historic mean of 12 points.

Decision: Fail to reject H₀.
